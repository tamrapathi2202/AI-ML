```python
#35.
import pandas as pd
contributions = [6700, 8000, 5400, 3400]
sections = ['D', 'A', 'B', 'C']
s12 = pd.Series(data=contributions, index=sections)
print("Series sorted by values (descending):")
print(s12.sort_values(ascending=False))
print("\nSeries sorted by indices (ascending):")
print(s12.sort_index())
```
