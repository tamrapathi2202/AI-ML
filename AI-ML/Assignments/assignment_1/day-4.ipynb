1. A company recorded daily sales for 7 days as [120, 135, 150, 160, 
145, 170, 155]. 
○ Write a program to compute the total and average sales.

2.Write a Python program that accepts a list of numbers from the user 
and prints a new list containing only the even numbers using a list 
comprehension.

```python
numbers = input("Enter numbers separated by commas: ") 
num_list = [int(num) for num in numbers.split(",")] # Convert input strings to integers
even_numbers = [num for num in num_list if num % 2 == 0]  # List comprehension to get even numbers
print("Even numbers:", even_numbers)         
```

4. Create a program that counts how many times each character 
occurs in the string "python programming". Use both: 

○ a regular dictionary with get()
 
○ a defaultdict(int)

```python
# Count character occurrences using a regular dictionary with get()
text = "python programming"
char_count = {}
for char in text:
    char_count[char] = char_count.get(char, 0) + 1
print("Using regular dict with get():", char_count)

# Count character occurrences using defaultdict(int)
from collections import defaultdict
char_count_dd = defaultdict(int)
for char in text:
    char_count_dd[char] += 1
print("Using defaultdict(int):", dict(char_count_dd))

```

7.Write a function factorial(n) that uses a loop to calculate factorial. 
Then write another version using recursion. Compare the results. 


```python

def factorial(n):
    result = 1
    for i in range(1,n+1):
        result *= i
    return result
n=int(input("Enter a number: "))
print(factorial(n))


#Then write another version using recursion. Compare the results.

def factorial_recur(numb):
    if numb == 0 or numb == 1:
        return 1
    else:
        return numb * factorial_recur(numb - 1)
numb=int(input("Enter a number: "))
print(factorial_recur(numb))
```

9. Write a closure function make_counter() that returns a function 
increment() which increases a count by 1 every time it is called. 
Demonstrate it by calling 5 times.

```python

def make_counter():
    count = 0
    def increment():
        nonlocal count   
        count += 1
        return count
    return increment

make_counter()

for _ in range(5):
    print(make_counter())
```

12. Write a Python program that: 

● Creates a list of squares from 1–10 using a list comprehension. 

● Creates a dictionary mapping numbers 1–10 to their cubes using a 
dictionary comprehension. 

● Creates a set of all remainders when numbers 1–20 are divided by 5 
using a set comprehension. 

```python
# 1. List of squares from 1–10 using list comprehension
squares = [x**2 for x in range(1, 11)]
print("List of squares:", squares)

# 2. Dictionary mapping numbers 1–10 to their cubes using dictionary comprehension
cubes = {x: x**3 for x in range(1, 11)}
print("Dictionary of cubes:", cubes)

# 3. Set of all remainders when numbers 1–20 are divided by 5 using set comprehension
remainders = {x % 5 for x in range(1, 21)}
print("Set of remainders:", remainders)
```

13. Given the dictionary: 
d = {"a": 1, "b": 2, "c": 3} 
Try deleting keys while iterating. Catch the error and then 
show the correct way using list(d.keys()).

```python
#error 
d = {"a": 1, "b": 2, "c": 3}
try:
    for key in d:
        del d[key]
except RuntimeError as e:
    print("Error:", e)
```

```python
#correct way
d = {"a": 1, "b": 2, "c": 3}

for key in list(d.keys()):   # make a copy of keys
    if d[key] % 2 == 1:      # if value is odd
        del d[key]           # safe to delete

print("Updated dictionary:", d)
```

14.Write a Python program to flatten this nested list using a nested 
loop comprehension: 
matrix = [[1, 2], [3, 4, 5], [6]] 
Expected Output: [1, 2, 3, 4, 5, 6] 

```python
matrix = [[1, 2], [3, 4, 5], [6]]

# Flatten using nested loop comprehension
flattened = [num for row in matrix for num in row]

print("Flattened list:", flattened)
```

15.Create a program to find the most frequent word in a sentence 
entered by the user. Print the word and its frequency. 

```python
sentence = input("Enter a sentence:")

words = sentence.split()
most_word = max(words, key=words.count)
frequency = words.count(most_word)

print("Most frequent word:", most_word,"Frequency:", frequency)
```
