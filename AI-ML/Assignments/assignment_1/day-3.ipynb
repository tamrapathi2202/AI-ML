1.In Quiz 1 of a course the marks obtained by ten students are 12, 4, 5, 3, 8, 7, 10, 15, 9, 11. Write a python program to calculate the mean and median of the marks obtained by the students. Cross check the results by importing statistics module in python.

```python
marks=[12, 4, 5, 3, 8, 7, 10, 15, 9, 11]
mean=sum(marks)/len(marks)     #mean
marks.sort()
n = len(marks)
median = (marks[n//2 - 1] + marks[n//2]) / 2  # since n=10 (even)    #median

print("Manual Mean:", mean)   
print("Manual Median:", median)
         
# Cross-check
import statistics
print("Statistics Mean:", statistics.mean(marks))
print("Statistics Median:", statistics.median(marks))

```

2.Find the second largest number in a list without using sort().

```python
numbers = [12, 4, 5, 3, 8, 7, 10, 15, 9, 11]

largest = max(numbers)        # find the largest
numbers.remove(largest)       # remove it once   #12 will be removed
second_largest = max(numbers) # now max of remaining is 2nd largest

print("Second Largest Number:", second_largest)
```
