2.. Suppose you’re designing an E-commerce app. Describe how you would use: 
○ Encapsulation 
○ Inheritance 
○ Polymorphism 
○ Abstraction 
Give class examples for each. 

```python
class Product:
    def __init__(self, name, price, stock):
        self.__name = name      # private
        self.__price = price    # private
        self.__stock = stock    # private

    def get_price(self):
        return self.__price
    
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price

    def buy(self, qty):
        if qty <= self.__stock:
            self.__stock -= qty
            print(f"✅ Bought {qty} {self.__name}(s)")
        else:
            print("❌ Not enough stock!")
```

```python
class Electronic(Product):   # inherits Product
    def __init__(self, name, price, stock, warranty):
        super().__init__(name, price, stock)
        self.warranty = warranty

class Clothing(Product):     # inherits Product
    def __init__(self, name, price, stock, size):
        super().__init__(name, price, stock)
        self.size = size
```

```python
class Electronic(Product):
    def __init__(self, name, price, stock, warranty):
        super().__init__(name, price, stock)
        self.warranty = warranty

    def get_details(self):   # overridden method
        return f"Electronic: {self.warranty} yr warranty, Price: {self.get_price()}"

class Clothing(Product):
    def __init__(self, name, price, stock, size):
        super().__init__(name, price, stock)
        self.size = size

    def get_details(self):   # overridden method
        return f"Clothing: Size {self.size}, Price: {self.get_price()}"


# Polymorphism in action
items = [
    Electronic("Laptop", 60000, 10, 2),
    Clothing("T-Shirt", 500, 50, "M")
]

for item in items:
    print(item.get_details())
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):   # abstract class
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f" Paid {amount} using Credit Card")

class UpiPayment(Payment):
    def pay(self, amount):
        print(f" Paid {amount} using UPI")


# Abstraction in action
p1 = CreditCardPayment()
p1.pay(1500)

p2 = UpiPayment()
p2.pay(700)
```

5. Compare composition vs inheritance. Give one scenario where you would 
prefer composition over inheritance. 

Inheritance → "is-a" relationship (e.g., Dog is an Animal).

Composition → "has-a" relationship (e.g., Car has an Engine).

Prefer composition when a class needs functionality from another class without being its subtype.
Example: A Library class uses (has) many Book objects. Here, composition is better than inheritance because a library is not a book, it just contains books.


```python
# --- Inheritance Example ---
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):   # Dog IS-A Animal
    def speak(self):
        return "Bark!"

dog = Dog()
print("Inheritance:", dog.speak())


# --- Composition Example ---
class Engine:
    def start(self):
        return "Engine started"

class Car:           # Car HAS-A Engine
    def __init__(self):
        self.engine = Engine()

    def drive(self):
        return self.engine.start() + " and Car is moving"

car = Car()
print("Composition:", car.drive())
```

6.Create a class hierarchy for Employees: 
○ Base class Employee (name, salary). 
○ Subclass Manager (manages list of employees). 
○ Subclass Developer (programming_language). 
Demonstrate polymorphism by calling a work() method for different 
employees.

```python
# Base Class
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def work(self):
        return f"{self.name} is working."


# Subclass: Manager
class Manager(Employee):
    def __init__(self, name, salary):
        super().__init__(name, salary)
        self.employees = []  # list of employees managed

    def add_employee(self, employee):
        self.employees.append(employee)

    def work(self):
        return f"{self.name} is managing {len(self.employees)} employees."


# Subclass: Developer
class Developer(Employee):
    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def work(self):
        return f"{self.name} is coding in {self.programming_language}."


# Demonstration of Polymorphism
if __name__ == "__main__":
    emp1 = Employee("John", 50000)
    dev1 = Developer("Alice", 70000, "Python")
    dev2 = Developer("Bob", 65000, "Java")
    mgr1 = Manager("Eve", 90000)

    mgr1.add_employee(dev1)
    mgr1.add_employee(dev2)

    # List of employees (different types)
    employees = [emp1, dev1, dev2, mgr1]

    # Polymorphic behavior
    for emp in employees:
        print(emp.work())
```

7.Implement multiple inheritance in Python with a Teacher and Researcher 
class. Derive a Professor class that inherits from both. Show how MRO 
(Method Resolution Order) works in Python.

```python
class Teacher:
    def teach(self):
        return "Teaching students"

class Researcher:
    def research(self):
        return "Doing research"

class Professor(Teacher, Researcher):
    def work(self):
        return "Managing both teaching and research"

# Create object
prof = Professor()
print(prof.teach())
print(prof.research())
print(prof.work())

# Show Method Resolution Order
print(Professor.__mro__)
```

8.. Design a Student Grading System using OOP: 
○ Class Student with attributes name, roll, marks (dict). 
○ Methods: calculate_average(), get_grade(). 
○ Use encapsulation to keep marks private. 

```python
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.__marks = marks   # private

    def calculate_average(self):
        return sum(self.__marks.values()) / len(self.__marks)

    def get_grade(self):
        avg = self.calculate_average()
        if avg >= 90:return 'A'
        elif avg >= 80:return 'B'
        elif avg >= 70:return 'C'
        elif avg >= 60:return 'D'
        else:return 'F'

# Example
s1 = Student("Alice", 101, {"Math": 95, "Science": 88, "English": 92})
print("Average:", s1.calculate_average())
print("Grade:", s1.get_grade())
```

9. Create an abstract class Payment with abstract method pay(amount). 
Implement subclasses CreditCardPayment, UPIPayment, and 
WalletPayment. Simulate different payments.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        return f"Paid {amount} using Credit Card"

class UPIPayment(Payment):
    def pay(self, amount):
        return f"Paid {amount} using UPI"

class WalletPayment(Payment):
    def pay(self, amount):
        return f"Paid {amount} using Wallet"

# Simulate
payments = [CreditCardPayment(), UPIPayment(), WalletPayment()]

for p in payments:
    print(p.pay(500))
```

10.. Define a custom exception InsufficientFundsError. Modify your 
BankAccount class so that withdrawing more than balance raises this 
exception. Handle it gracefully.

```python
class InsufficientFundsError(Exception):
    pass

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError("Insufficient funds in account")
        self.balance -= amount
        return f"Withdrawn {amount}, Remaining balance: {self.balance}"

# Example
acc = BankAccount(1000)

try:
    print(acc.withdraw(500))
    print(acc.withdraw(600))  # exceeds balance
except InsufficientFundsError as e:
    print("Error:", e)
```

12. Implement a library management system: 
● Class Book with attributes: title, author, copies. 
● Class Library that manages a collection of books with methods: add_book, 
borrow_book, return_book. 
● Use encapsulation to prevent direct modification of copies. 

```python
class Book:
    def __init__(self, title, author, copies):
        self.title = title
        self.author = author
        self.__copies = copies   # private

    def borrow(self):
        if self.__copies > 0:
            self.__copies -= 1
            return True
        return False

    def return_book(self):
        self.__copies += 1

    def get_copies(self):
        return self.__copies


class Library:
    def __init__(self):
        self.books = {}

    def add_book(self, book):
        self.books[book.title] = book

    def borrow_book(self, title):
        if title in self.books and self.books[title].borrow():
            print(f"Borrowed: {title}")
        else:
            print(f"{title} not available")

    def return_book(self, title):
        if title in self.books:
            self.books[title].return_book()
            print(f"Returned: {title}")


# Example
lib = Library()
lib.add_book(Book("Python", "Guido", 2))
lib.borrow_book("Python")
lib.borrow_book("Python")
lib.borrow_book("Python")
lib.return_book("Python")
```

15. Build a Shape hierarchy where: 
● Shape is an abstract class with abstract methods area() and perimeter(). 
● Implement Rectangle, Circle, and Triangle. 
● Raise a custom exception InvalidShapeError if negative dimensions are 
given. 

```python
from abc import ABC, abstractmethod
import math

class InvalidShapeError(Exception): pass

class Shape(ABC):
    @abstractmethod
    def area(self): pass

    @abstractmethod
    def perimeter(self): pass

class Rectangle(Shape):
    def __init__(self, w, h):
        if w < 0 or h < 0: raise InvalidShapeError("Negative dimensions not allowed")
        self.w, self.h = w, h
    def area(self): return self.w * self.h
    def perimeter(self): return 2 * (self.w + self.h)

class Circle(Shape):
    def __init__(self, r):
        if r < 0: raise InvalidShapeError("Negative radius not allowed")
        self.r = r
    def area(self): return math.pi * self.r**2
    def perimeter(self): return 2 * math.pi * self.r

class Triangle(Shape):
    def __init__(self, a, b, c):
        if a < 0 or b < 0 or c < 0: raise InvalidShapeError("Negative sides not allowed")
        self.a, self.b, self.c = a, b, c
    def area(self):
        s = (self.a + self.b + self.c) / 2
        return math.sqrt(s*(s-self.a)*(s-self.b)*(s-self.c))
    def perimeter(self): return self.a + self.b + self.c

# Example
shapes = [Rectangle(4, 5), Circle(3), Triangle(3, 4, 5)]
for s in shapes:
    print(type(s).__name__, "Area:", s.area(), "Perimeter:", s.perimeter())
```
