```python
#1.
import matplotlib.pyplot as plt
years = [1951, 1961, 1971, 1981, 1991, 2001, 2011]
population = [361088000, 439235000, 548160000, 683329000, 
              846387888, 1028737436, 1210726932]
plt.figure(figsize=(10,6))
plt.plot(years, population, marker='o', color='blue', linewidth=2)
plt.xlabel("Census Year", fontsize=12)
plt.ylabel("Population", fontsize=12)
plt.title("Population Growth in India (1951–2011)", fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.7)
for x, y in zip(years, population):
    plt.text(x, y+20000000, f"{y:,}", ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
#2.
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 500)   # 500 points between 0 and 2π
y = np.sin(x)

plt.figure(figsize=(10,6))
plt.plot(x, y, color='blue', linewidth=2, label='y = sin(x)')
plt.xlabel("x (radians)", fontsize=12)
plt.ylabel("sin(x)", fontsize=12)
plt.title("Sine Curve", fontsize=14, fontweight='bold')

# Add grid and legend
plt.axhline(0, color='black', linewidth=0.8)  # x-axis
plt.axvline(0, color='black', linewidth=0.8)  # y-axis
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#3.
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 500)   # 500 points between 0 and 2π
y = np.cos(x)

# Plot sine curve
plt.figure(figsize=(10,6))
plt.plot(x, y, color='blue', linewidth=2, label='y = cos(x)')

# Labels and title
plt.xlabel("x (radians)", fontsize=12)
plt.ylabel("cos(x)", fontsize=12)
plt.title("cosine Curve", fontsize=14, fontweight='bold')

# Add grid and legend
plt.axhline(0, color='black', linewidth=0.8)  # x-axis
plt.axvline(0, color='black', linewidth=0.8)  # y-axis
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#4. Plot a Bar Graph to show the number of boys in each class 6- 12. Data should be imagined
# by student.
import matplotlib.pyplot as plt
classes = ["6", "7", "8", "9", "10", "11", "12"]
boys = [35, 42, 38, 40, 45, 37, 32]

# Plot bar graph
plt.figure(figsize=(8,6))
plt.bar(classes, boys, color='red', edgecolor='black')
plt.xlabel("Class", fontsize=12)
plt.ylabel("Number of Boys", fontsize=12)
plt.title("Number of Boys in Classes 6 to 12", fontsize=14, fontweight='bold')

# Add values on top of bars
for i, val in enumerate(boys):
    plt.text(i, val + 1, str(val), ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
#5. Plot a Bar Graph for Marks scored in different subjects. Data should be imagined.
import matplotlib.pyplot as plt
subjects = ["Math", "Science", "English", "History", "Computer"]
marks = [88, 92, 75, 68, 95]
plt.figure(figsize=(8,6))
plt.bar(subjects, marks, color='green', edgecolor='black')
plt.xlabel("Subjects", fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.title("Marks Scored in Different Subjects", fontsize=14, fontweight='bold')
plt.ylim(0, 100)  # since marks are out of 100
plt.tight_layout()
plt.show()

```

```python
#6. PlotaHistogramtofindthenumberofemployeescomingtoofficebetween7amto12noon.
# Use bins as 1 hr gaps.
import matplotlib.pyplot as plt
arrival_times = [7.2, 7.5, 7.8, 8.0, 8.3, 8.7, 9.0, 9.1, 9.4, 9.8,
                 10.0, 10.2, 10.5, 11.0, 11.2, 11.5, 11.8, 7.4, 8.1, 9.6]
bins = [7, 8, 9, 10, 11, 12]
plt.figure(figsize=(8,6))
plt.hist(arrival_times, bins=bins, color='pink', edgecolor='black', rwidth=0.9)
plt.xlabel("Time of Arrival (Hours)", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)
plt.title("Employee Arrivals Between 7 AM to 12 Noon", fontsize=14, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
